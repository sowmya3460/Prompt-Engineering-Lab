```python
#experiment3:iterative refinement on a sample task:summarize the plot of the shakepearean play remeo and juliet in two sentences through three rounds of prompt tweaking
```

```python
from google import genai

# Replace with your Gemini API Key
API_KEY = GOOGLE_API_KEY

# Create client
client = genai.Client(api_key=API_KEY)

# List of prompts
prompts = [
    # Iteration 1
    "Summarize the plot of Shakespeare's play Romeo and Juliet in two sentences.",

    # Iteration 2
    "Summarize the plot of Shakespeare's play Romeo and Juliet in exactly two concise and formal sentences suitable for high school literature students.",

    # Iteration 3
    "Summarize the plot of Shakespeare's play Romeo and Juliet in exactly two formal sentences. Include the setting (Verona, Italy) and emphasize the themes of love, fate, and family conflict."
]

# Generate responses
for i, prompt in enumerate(prompts, start=1):
    print("=" * 70)
    print(f"Iteration {i}")
    print("=" * 70)
    print("\nPrompt:")
    print(prompt)

    response = client.models.generate_content(
        model="gemini-flash-latest",
        contents=prompt
    )

    print("\nGenerated Summary:")
    print(response.text)
    print("\n")
```

```python
# Different prompt versions
prompts = [
    {
        "title": "1. Vague Prompt",
        "prompt": "Tell me about Python."
    },
    {
        "title": "2. Contradictory Prompt",
        "prompt": "Explain Python in one sentence and provide a detailed explanation with examples."
    },
    {
        "title": "3. Refined Prompt",
        "prompt": """
Explain Python programming for beginners.

Instructions:
- Write exactly five bullet points.
- Keep each point under 20 words.
- Mention:
    * What Python is
    * Major features
    * Common applications
    * Advantages
    * One real-world example

Example format:
• Python is an easy-to-learn programming language.
"""
    }
]

# Generate responses
for item in prompts:
    print("="*80)
    print(item["title"])
    print("="*80)

    print("\nPrompt:")
    print(item["prompt"])

    response = client.models.generate_content(
        model="gemini-flash-latest",
        contents=item["prompt"]
    )

    print("\nGenerated Response:")
    print(response.text)
    print("\n")
```
