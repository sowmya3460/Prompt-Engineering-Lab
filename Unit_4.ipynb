```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import google.generativeai as genai
from IPython.display import Markdown
from google.colab import userdata
GOOGLE_API_KEY=userdata.get('APIKEY1')
genai.configure(api_key=GOOGLE_API_KEY)
import google.generativeai as genai
model = genai.GenerativeModel("gemini-flash-latest")

GOOGLE_API_KEY = GOOGLE_API_KEY

llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    google_api_key=GOOGLE_API_KEY
)
```

```python
prompt = ChatPromptTemplate.from_template(
"""
Summarize the following text in 50 words.

Text:
{text}
"""
)

parser = StrOutputParser()

chain = prompt | llm | parser

text = """
Artificial Intelligence enables computers to perform tasks that normally require human intelligence,
such as learning, reasoning, decision making, and language understanding.
"""

result = chain.invoke({"text": text})

print(result)
```

```python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python

# ------------------------------------
# Step 1: Create Sample Documents
# ------------------------------------

docs = [
    Document(page_content="Artificial Intelligence enables machines to learn from data."),

    Document(page_content="Machine Learning is a subset of Artificial Intelligence."),

    Document(page_content="Deep Learning uses multi-layer neural networks."),

    Document(page_content="Natural Language Processing enables computers to understand human language.")
]
```

```python
# ------------------------------------
# Step 2: Split Documents
# ------------------------------------

splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

split_docs = splitter.split_documents(docs)

print("Number of Chunks:", len(split_docs))

# ------------------------------------
# Step 3: Load Embedding Model
# ------------------------------------

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# ------------------------------------
# Step 4: Create FAISS Vector Store
# ------------------------------------

db = FAISS.from_documents(
    split_docs,
    embeddings
)

print("Vector Store Created Successfully")

# ------------------------------------
# Step 5: Similarity Search
# ------------------------------------

query = "What is Machine Learning?"

results = db.similarity_search(
    query,
    k=2
)

print("\nTop Matching Documents\n")

for i, doc in enumerate(results, 1):
    print(f"Result {i}")
    print(doc.page_content)
    print("-" * 50)
```

```python
#Build a rag pipeline
#a.Recieves a user query
#b.Retrieves the top-k relevant chunks
#c.constructs acombined prompt with context+query
#d.send it to the LLM
#e.Returns the answer
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import ChatPromptTemplate
from google.colab import userdata
import os

# Install langchain-google-genai if not already installed
!pip install langchain-google-genai

# -------------------------------------------------
# STEP 1 : Gemini API Key
# -------------------------------------------------
GOOGLE_API_KEY = userdata.get("APIKEY1")
os.environ["GOOGLE_API_KEY"] = GOOGLE_API_KEY # Ensure env var is also set

llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    google_api_key=GOOGLE_API_KEY,
    temperature=0
)
```

```python
documents = [

Document(page_content="""
Artificial Intelligence (AI) is the science of building intelligent systems.
Machine Learning is a subset of AI.
Deep Learning is a subset of Machine Learning.
"""),

Document(page_content="""
Natural Language Processing (NLP) enables computers
to understand and generate human language.
Applications include chatbots, translation and sentiment analysis.
"""),

Document(page_content="""
Computer Vision enables machines to recognize images,
objects and videos.
Applications include face recognition and autonomous vehicles.
""")
]

# -------------------------------------------------
# STEP 3 : Split Documents
# -------------------------------------------------

splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

chunks = splitter.split_documents(documents)

print("Chunks:", len(chunks))

# -------------------------------------------------
# STEP 4 : Embeddings
# -------------------------------------------------

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# -------------------------------------------------
# STEP 5 : Build FAISS Index
# -------------------------------------------------

vectorstore = FAISS.from_documents(
    chunks,
    embedding_model
)

print("FAISS Index Created")

# -------------------------------------------------
# STEP 6 : Retriever
# -------------------------------------------------

retriever = vectorstore.as_retriever(
    search_kwargs={"k":2}
)

# -------------------------------------------------
# STEP 7 : Gemini LLM
# -------------------------------------------------

llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    temperature=0
)

# -------------------------------------------------
# STEP 8 : Prompt Template
# -------------------------------------------------

prompt = ChatPromptTemplate.from_template("""

Answer the question only using the context below.

Context:
{context}

Question:
{question}

Answer:

""")

# -------------------------------------------------
# STEP 9 : User Query
# -------------------------------------------------

query = "What is Machine Learning?"

# -------------------------------------------------
# STEP 10 : Retrieve Top-K Documents
# -------------------------------------------------

retrieved_docs = retriever.invoke(query)

context = "\n\n".join(
    doc.page_content
    for doc in retrieved_docs
)

print("\nRetrieved Context\n")
print(context)

# -------------------------------------------------
# STEP 11 : Construct Prompt
# -------------------------------------------------

final_prompt = prompt.invoke(
    {
        "context":context,
        "question":query
    }
)

# -------------------------------------------------
# STEP 12 : Send to Gemini
# -------------------------------------------------

response = llm.invoke(final_prompt)

print("\nRAG Answer\n")
print(response.content)

# -------------------------------------------------
# STEP 13 : Compare WITHOUT Retrieval
# -------------------------------------------------

print("\nWithout Retrieval\n")

plain_response = llm.invoke(query)

print(plain_response.content)

```
